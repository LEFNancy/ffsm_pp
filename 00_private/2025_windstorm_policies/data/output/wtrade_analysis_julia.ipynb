```julia
using DataArrays, DataFrames

scenarios = ["wtrade_const", "wtrade10", "wtrade08", "wtrade09", "wtrade11", "wtrade12"]
baseIDir = "results";
productData_byscenario = [readtable(baseIDir*"/productData_"*scenario*".csv", separator = ';') for scenario in scenarios];


```

```julia
prodVars = vcat(productData_byscenario)
[delete!(dfi, [:x,:country,:freeDim]) for dfi in [prodVars]]; # delete columns X (empty at the end) and country
head(prodVars)
```
