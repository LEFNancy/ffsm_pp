```julia
using DataArrays, DataFrames

scenarios = ["debug","upper_limited"]
baseIDir = "results";
forestData_byscenario = [readtable(baseIDir*"/forestData_"*scenario*".csv", separator = ';') for scenario in scenarios];

```

```julia
forVars = vcat(forestData_byscenario)
[delete!(dfi, [:x,:country,:freeDim,:region,:forType]) for dfi in [forVars]]; # delete columns X (empty at the end) and country
head(forVars)
```

```julia
forVars = aggregate(forVars, [:parName,:scen, :year], sum)
#aggregate(forestDataReg, :region, sum)
```
